```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
X, y_true = make_blobs(n_samples=500, centers=3, cluster_std=4, random_state=42)

```

```python
X
```

```python
df=pd.DataFrame(X,columns=['Feature_1','Feature_2'])
```

```python
df
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
#here inertia is WCSS
```

```python
inertia =[]
K_range = range(1,11)
```

```python


for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
inertia
```

```python
plt.plot(K_range,inertia,marker ='o')
```

```python
kmeans_final = KMeans(n_clusters=4,random_state=42)
```

```python


cluster_labels = kmeans_final.fit_predict(X_scaled)
```

```python
cluster_labels
```

```python
df['cluster']= cluster_labels
```

```python
sns.scatterplot(x=df['Feature_1'],
                y=df['Feature_2'],
                hue=df['cluster'],
                palette='viridis')
```

```python
import seaborn as sns
import pandas as pd
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Regenerate make_blobs data for KMeans plot context
X_blobs, y_true_blobs = make_blobs(n_samples=500, centers=3, cluster_std=4, random_state=42)
df_blobs = pd.DataFrame(X_blobs, columns=['Feature_1', 'Feature_2'])

# Scale data
scaler_blobs = StandardScaler()
X_scaled_blobs = scaler_blobs.fit_transform(df_blobs)

# Apply KMeans
kmeans_final_blobs = KMeans(n_clusters=4, random_state=42, n_init='auto') # Added n_init='auto' for newer sklearn versions
cluster_labels_blobs = kmeans_final_blobs.fit_predict(X_scaled_blobs)

# Add cluster labels to DataFrame
df_blobs['kmeans_cluster'] = cluster_labels_blobs

# Plot KMeans clusters
sns.scatterplot(x=df_blobs['Feature_1'], y=df_blobs['Feature_2'], hue=df_blobs['kmeans_cluster'], palette='tab10')
```

```python
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_moons
import pandas as pd

# Generate make_moons data
X, y_true = make_moons(n_samples=500, noise=0.05, random_state=42)
# Create DataFrame
df = pd.DataFrame(X, columns=['Feature_1', 'Feature_2'])

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)

# Apply DBSCAN
dbscan = DBSCAN(eps=0.3, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster']= dbscan_labels
```

```python
sns.scatterplot(x=df['Feature_1'], y=df['Feature_2'], hue=df['dbscan_cluster'], palette='tab10')

```
